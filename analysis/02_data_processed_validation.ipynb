## Pemeriksaan Ringkasan File CSV (Rows, Cols, Null)

```python
from pathlib import Path
import pandas as pd

# 1. Mendapatkan direktori utama proyek
PROJECT_ROOT = Path.cwd().parent

# 2. Menentukan path ke folder data processed
processed_dir = PROJECT_ROOT / "data" / "processed"

# 3. Pemetaan nama entitas ke file CSV hasil ETL
files = {
    "categories": "categories.csv",
    "products": "products.csv",
    "stores": "stores.csv",
    "sales": "sales_transactions.csv",
}

print("=== VERIFIKASI RINGKASAN DATA PROCESSED ===")

# 4. Iterasi untuk membaca dan mengecek ringkasan setiap file CSV
for name, filename in files.items():
    path = processed_dir / filename
    df_check = pd.read_csv(path)

    print(f"\n[ TABEL: {name.upper()} ]")
    print("Jumlah Baris (Rows) :", len(df_check))
    print("Jumlah Kolom (Cols) :", len(df_check.columns))
    print("Total Missing Values (Null) :", df_check.isna().sum().sum())
    print("-" * 40)
```

## Validasi Integritas Referensial (Foreign Key Checking)

```python
# 1. Membaca kembali ke-4 file CSV hasil ETL ke dalam DataFrame
categories = pd.read_csv(processed_dir / "categories.csv")
products = pd.read_csv(processed_dir / "products.csv")
stores = pd.read_csv(processed_dir / "stores.csv")
sales = pd.read_csv(processed_dir / "sales_transactions.csv")

# 2. Membuat kumpulan (set) ID yang valid dari tabel-tabel master
valid_category_ids = set(categories["category_id"])
valid_product_ids = set(products["product_id"])
valid_store_ids = set(stores["store_id"])

# 3. Pengecekan FK 1: Apakah ada category_id di tabel products yang tidak ada di tabel categories?
invalid_product_categories = products.loc[
    ~products["category_id"].isin(valid_category_ids)
]

# 4. Pengecekan FK 2: Apakah ada product_id di tabel sales yang tidak ada di tabel products?
invalid_sales_products = sales.loc[
    ~sales["product_id"].isin(valid_product_ids)
]

# 5. Pengecekan FK 3: Apakah ada store_id di tabel sales yang tidak ada di tabel stores?
invalid_sales_stores = sales.loc[
    ~sales["store_id"].isin(valid_store_ids)
]

# 6. Menampilkan hasil pengujian integritas Foreign Key
print("=== HASIL VALIDASI RELEVANSI FOREIGN KEY (FK) ===")
print("Invalid product → category :", len(invalid_product_categories))
print("Invalid sales → product   :", len(invalid_sales_products))
print("Invalid sales → store     :", len(invalid_sales_stores))
```
