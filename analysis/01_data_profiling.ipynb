## Import Library & Inspeksi Gambaran Umum Dataset

```python
import pandas as pd

# 1. Membaca dataset dari file Excel
df = pd.read_excel("../data/raw/coffee_shop_sales.xlsx")

# 2. Menampilkan dimensi/ukuran dataset
print("=== DIMENSI DATASET (Baris, Kolom) ===")
print("Shape:", df.shape)

# 3. Menampilkan daftar seluruh nama kolom
print("\n=== DAFTAR KOLOM ===")
print(df.columns.tolist())

# 4. Menampilkan tipe data tiap kolom
print("\n=== TIPE DATA SETIAP KOLOM ===")
print(df.dtypes)

# 5. Cek jumlah nilai kosong (missing values) di setiap kolom
print("\n=== JUMLAH MISSING VALUES PER KOLOM ===")
print(df.isna().sum())

# 6. Cek total baris yang duplikat
print("\n=== JUMLAH BARIS DUPLIKAT ===")
print("Total Duplikat:", df.duplicated().sum())

# 7. Menghitung jumlah nilai unik per kolom
print("\n=== JUMLAH NILAI UNIK PER KOLOM ===")
print(df.nunique())

# 8. Menampilkan 5 baris pertama sampel data
print("\n=== PRATINJAU 5 BARIS PERTAMA (PREVIEW) ===")
display(df.head())
```

## Detail Exploratory Data Analysis (EDA)

```python
# 1. Informasi Ukuran Dataset
print("=== INFORMASI UKURAN DATASET ===")
print("Jumlah baris :", len(df))
print("Jumlah kolom :", len(df.columns))

# 2. Rentang Tanggal Transaksi
print("\n=== RENTANG PERIODE TRANSAKSI ===")
print("Start :", df["transaction_date"].min())
print("End   :", df["transaction_date"].max())

# 3. Cek Missing Values (Diurutkan dari yang terbanyak)
print("\n=== JUMLAH MISSING VALUES (DARI TERBANYAK) ===")
missing = (
    df.isna()
      .sum()
      .sort_values(ascending=False)
)
print(missing)

# 4. Cek Data Duplikat (Baris secara keseluruhan vs ID Transaksi)
print("\n=== PENGECEKAN DATA DUPLIKAT ===")
print("Duplicate rows          :", df.duplicated().sum())
print("Duplicate transaction_id:", df["transaction_id"].duplicated().sum())

# 5. Jumlah Nilai Unik per Kolom Kategori & Lokasi
print("\n=== JUMLAH NILAI UNIK KATEGORI & LOKASI ===")
for col in [
    "store_id",
    "store_location",
    "product_id",
    "product_category",
    "product_type",
    "product_detail"
]:
    print(f"{col}: {df[col].nunique()} unique")

# 6. Statistik Deskriptif untuk Variabel Numerik
print("\n=== STATISTIK DESKRIPTIF: TRANSACTION QUANTITY ===")
print(df["transaction_qty"].describe())

print("\n=== STATISTIK DESKRIPTIF: UNIT PRICE ===")
print(df["unit_price"].describe())

# 7. Daftar Referensi Toko Unik
print("\n=== DAFTAR TOKO DAN LOKASI (UNIQUE) ===")
print(
    df[["store_id", "store_location"]]
    .drop_duplicates()
    .sort_values("store_id")
    .to_string(index=False)
)

# 8. Daftar Referensi Produk Unik beserta Harganya
print("\n=== DAFTAR PRODUK DAN HARGA (UNIQUE) ===")
print(
    df[
        [
            "product_id",
            "product_category",
            "product_type",
            "product_detail",
            "unit_price"
        ]
    ]
    .drop_duplicates()
    .sort_values("product_id")
    .to_string(index=False)
)
```

## Validasi Konsistensi Harga Produk

```python
# 1. Menghitung berapa banyak harga unik yang dimiliki oleh setiap product_id
print("=== JUMLAH VARIASI HARGA PER PRODUCT_ID ===")
price_check = (
    df.groupby("product_id")["unit_price"]
      .nunique()
      .sort_values(ascending=False)
)
print(price_check)

# 2. Menampilkan produk yang memiliki harga tidak konsisten (lebih dari 1 variasi harga)
print("\n=== DAFTAR PRODUK DENGAN LEBIH DARI 1 HARGA (ANOMALI) ===")
inconsistent_prices = price_check[price_check > 1]

if inconsistent_prices.empty:
    print("Tidak ada anomali. Semua produk memiliki harga yang konsisten.")
else:
    print(inconsistent_prices)
```

## Perhitungan Revenue (Pendapatan)

```python
# 1. Menghitung revenue per transaksi (Jumlah Barang x Harga Satuan)
df["revenue"] = df["transaction_qty"] * df["unit_price"]

# 2. Statistik deskriptif revenue per transaksi
print("=== STATISTIK DESKRIPTIF REVENUE PER TRANSAKSI ===")
print(df["revenue"].describe())

# 3. Total revenue keseluruhan
print("\n=== TOTAL REVENUE KESELURUHAN ===")
print(f"Total Revenue: {df['revenue'].sum():,.2f}")

# 4. Total revenue yang dikelompokkan berdasarkan bulan
print("\n=== RINGKASAN TOTAL REVENUE PER BULAN ===")
monthly_revenue = (
    df.assign(month=df["transaction_date"].dt.to_period("M"))
      .groupby("month")["revenue"]
      .sum()
)
print(monthly_revenue)
```

## Agregasi Pendapatan Bulanan & Penjualan Harian

```python
# 1. Agregasi Total Revenue Bulanan
print("=== RINGKASAN TOTAL REVENUE PER BULAN ===")
monthly_revenue = (
    df.assign(month=df["transaction_date"].dt.to_period("M"))
      .groupby("month")["revenue"]
      .sum()
)
print(monthly_revenue)

# 2. Agregasi Penjualan Harian (Jumlah transaksi, Total item terjual, Total revenue)
daily_sales = (
    df.groupby("transaction_date")
      .agg(
          transactions=("transaction_id", "count"),
          quantity=("transaction_qty", "sum"),
          revenue=("revenue", "sum")
      )
)

print("\n=== PRATINJAU REKAP PENJUALAN HARIAN (5 HARI PERTAMA) ===")
print(daily_sales.head())
```

## Konsistensi Produk & Store (Data Quality Check)

```python
# Menghitung jumlah nilai unik dari kolom type, detail, dan category untuk setiap product_id
product_consistency = (
    df.groupby("product_id")
      .agg(
          product_type_nunique=("product_type", "nunique"),
          product_detail_nunique=("product_detail", "nunique"),
          category_nunique=("product_category", "nunique")
      )
)

print("=== PENGECEKAN ANOMALI / INKONSISTENSI DATA PRODUK ===")

# Menampilkan product_id yang memiliki lebih dari 1 tipe, detail, atau kategori (anomali)
print(product_consistency[
    (product_consistency["product_type_nunique"] > 1) |
    (product_consistency["product_detail_nunique"] > 1) |
    (product_consistency["category_nunique"] > 1)
])

# 2. PENGECEKAN KONSISTENSI LOKASI TOKO
# Menghitung berapa banyak nama store_location unik untuk setiap store_id
store_consistency = (
    df.groupby("store_id")["store_location"]
      .nunique()
)

print("\n=== PENGECEKAN ANOMALI / INKONSISTENSI LOKASI TOKO ===")

# Menampilkan store_id yang terikat ke lebih dari 1 lokasi toko (anomali)
print(
    store_consistency[store_consistency > 1]
)
```
